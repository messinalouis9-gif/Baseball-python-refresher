## 4. Python Basics In Five Minutes

**17. Store a value in a variable**  
Tip: Text goes in quotes; numbers do not. Names are case-sensitive: velo and Velo are different.

````python
velo = 95.4
pitcher = "Paul Skenes"
print(pitcher, velo)
````

**18. A list holds several values in order**  
Tip: Counting starts at 0, so fastballs[0] is the first item.

````python
fastballs = ["FF", "SI", "FC"]
print(fastballs[0], len(fastballs))
````

**19. A dictionary looks things up**  
Tip: Key on the left, value on the right. Like a two-column lookup table.

````python
pitch_names = {"FF": "4-Seam Fastball", "SL": "Slider", "CH": "Changeup"}
print(pitch_names["SL"])
````

**20. Make a decision with if**  
Tip: The colon and the indent (4 spaces) matter. Python uses indentation instead of brackets.

````python
if velo >= 95:
    print("Plus velocity")
else:
    print("Below 95")
````

**21. Repeat with a for loop**  
Tip: Runs the indented line once for each item in the list.

````python
for p in fastballs:
    print(p, "is a fastball code")
````

**22. Mix text and numbers with an f-string**  
Tip: :.1f rounds to one decimal place.

````python
gap = 95.4 - 87.1
print(f"Velo gap: {gap:.1f} mph")
````

## 8. Clean The Data

**40. Make dates real dates**  
Tip: Now you can filter by date and pull out months or weekdays.

````python
df["game_date"] = pd.to_datetime(df["game_date"])
````

**41. Convert Excel serial dates**  
Tip: Excel stores dates as day counts. 45550 becomes 2024-09-15.

````python
pd.to_datetime(45550, unit="D", origin="1899-12-30")
````

**42. Keep only the columns you need**  
Tip: .copy() makes a separate table, which avoids the SettingWithCopyWarning later.

````python
cols = ["game_pk", "game_date", "pitcher", "player_name", "p_throws", "stand",
        "inning", "inning_topbot", "home_team", "away_team", "at_bat_number",
        "pitch_number", "balls", "strikes", "pitch_type", "pitch_name",
        "release_speed", "release_spin_rate", "pfx_x", "pfx_z", "description",
        "events", "launch_speed", "launch_angle", "game_type"]
pitches = df[cols].copy()
````

**43. Drop rows missing key values**  
Tip: A few pitches each week have no pitch type or speed (tracking hiccups).

````python
pitches = pitches.dropna(subset=["pitch_type", "release_speed"])
````

**44. Remove exact duplicates**  
Tip: Always print how many rows a cleaning step removed.

````python
before = len(pitches)
pitches = pitches.drop_duplicates()
print(before - len(pitches), "duplicates removed")
````

**45. Rename a column**  
Tip: Clear names make your notebook easier to read later.

````python
pitches = pitches.rename(columns={"player_name": "pitcher_name"})
````

**46. Turn text into numbers**  
Tip: errors="coerce" turns anything that is not a number into a blank instead of crashing.

````python
pitches["release_speed"] = pd.to_numeric(pitches["release_speed"], errors="coerce")
````

## 9. Filter And Select

**47. Regular season only**  
Tip: Use == to test equals. One = assigns a value.

````python
pitches = pitches[pitches["game_type"] == "R"]
````

**48. One pitcher, by ID**  
Tip: Always filter by ID, not name.

````python
one = pitches[pitches["pitcher"] == 694973]
````

**49. Two conditions (and = &, or = |)**  
Tip: Wrap each condition in parentheses. Forgetting them is the most common error.

````python
lefty_sl = pitches[(pitches["p_throws"] == "L") & (pitches["pitch_type"] == "SL")]
````

**50. Match any value in a list**  
Tip: isin is cleaner than a chain of ors.

````python
FASTBALLS = ["FF", "SI", "FC"]
fb = pitches[pitches["pitch_type"].isin(FASTBALLS)]
````

**51. A date range**  
Tip: Works once game_date is a real date (section 8).

````python
early = pitches[pitches["game_date"].between("2024-05-11", "2024-05-13")]
````

**52. Which team is pitching?**  
Tip: Baseball logic in code: the home team pitches in the top of the inning. np.where works like Excel's IF.

````python
pitches["pitching_team"] = np.where(pitches["inning_topbot"] == "Top",
                                     pitches["home_team"], pitches["away_team"])
````

**53. Pick a few columns**  
Tip: Double brackets mean a list of columns.

````python
pitches[["pitcher_name", "pitching_team", "pitch_type", "release_speed"]].head()
````

## 10. Create New Columns

**54. Flag swings and whiffs**  
Tip: Sites define whiffs slightly differently (foul tips, bunts). Write your definition down and stick to it.

````python
SWINGS = ["swinging_strike", "swinging_strike_blocked", "foul", "foul_tip",
          "hit_into_play", "foul_bunt", "missed_bunt", "bunt_foul_tip"]
WHIFFS = ["swinging_strike", "swinging_strike_blocked"]
pitches["swing"] = pitches["description"].isin(SWINGS)
pitches["whiff"] = pitches["description"].isin(WHIFFS)
````

**55. Build the count**  
Tip: astype(str) turns numbers into text so they can be joined: 3-2, 0-1 and so on.

````python
pitches["count"] = pitches["balls"].astype(str) + "-" + pitches["strikes"].astype(str)
````

**56. Movement in inches, arm side positive**  
Tip: pfx_x is from the catcher's view, so righties' arm-side pitches start negative. Flipping them puts every pitcher on one scale: arm side positive, glove side negative.

````python
pitches["hb_in"] = pitches["pfx_x"] * 12
pitches["ivb_in"] = pitches["pfx_z"] * 12
# arm-side break is positive for every pitcher
pitches["hb_arm_side"] = np.where(pitches["p_throws"] == "R",
                                  -pitches["hb_in"],
                                  pitches["hb_in"])
````

**57. Group pitch codes with a lookup**  
Tip: map uses the dictionary from section 4. Codes not in it (knuckleball, eephus) become Other.

````python
GROUPS = {"FF": "Fastball", "SI": "Fastball", "FC": "Fastball",
          "SL": "Breaking", "ST": "Breaking", "CU": "Breaking",
          "KC": "Breaking", "SV": "Breaking",
          "CH": "Offspeed", "FS": "Offspeed"}
pitches["pitch_group"] = pitches["pitch_type"].map(GROUPS).fillna("Other")
````

**58. Number each pitcher's pitches in a game**  
Tip: Sort first so pitches are in the order they were thrown. Useful for fatigue questions.

````python
pitches = pitches.sort_values(["game_pk", "at_bat_number", "pitch_number"])
pitches["pitch_of_game"] = pitches.groupby(["game_pk", "pitcher"]).cumcount() + 1
````

## 11. Group And Summarize

**59. Average velo by pitcher and pitch type**  
Tip: reset_index() turns the groups back into normal columns.

````python
velo = (pitches.groupby(["pitcher", "pitcher_name", "pitch_type"])["release_speed"]
          .agg(n="count", avg_velo="mean").reset_index())
````

**60. Set a minimum sample**  
Tip: Small samples mislead. Pick a minimum before you look at the results.

````python
velo = velo[velo["n"] >= 30]
````

**61. Sort and show the top 10**  
Tip: ascending=True (the default) sorts smallest first.

````python
velo.sort_values("avg_velo", ascending=False).head(10)
````

**62. Whiff rate by pitch type**  
Tip: The mean of True/False is the share that are True, so mean() gives the whiff rate.

````python
sw = pitches[pitches["swing"]]
whiff_rate = sw.groupby("pitch_type")["whiff"].agg(swings="count", whiff_pct="mean")
whiff_rate.sort_values("whiff_pct", ascending=False)
````

**63. Pitch mix with a pivot table**  
Tip: Each row adds up to 1, so you read it as percent of that pitcher's pitches. Works like an Excel pivot table.

````python
mix = pd.crosstab(pitches["pitcher_name"], pitches["pitch_type"], normalize="index").round(3)
mix.loc[["Skenes, Paul"]]
````

**64. Attach a summary back to each row (merge)**  
Tip: merge is Python's VLOOKUP. how="left" keeps every pitch even if no match is found.

````python
totals = (pitches.groupby(["game_pk", "pitcher"]).size()
          .reset_index(name="pitches_in_game"))
pitches = pitches.merge(totals, on=["game_pk", "pitcher"], how="left")
````

## 13. Charts

**70. Histogram of 4-seam velocity**  
Tip: Always label axes. A chart without units is a guess.

````python
pitches.loc[pitches["pitch_type"] == "FF", "release_speed"].hist(bins=40)
plt.xlabel("Release speed (mph)")
plt.ylabel("Pitches")
plt.title("4-Seam Fastball Velocity")
plt.show()
````

**71. Bar chart of whiff rate**  
Tip: barh makes horizontal bars, easier to read with labels.

````python
whiff_rate["whiff_pct"].sort_values().plot(kind="barh")
plt.xlabel("Whiff rate on swings")
plt.show()
````

**72. Movement chart for one pitcher**  
Tip: One color per pitch type. Uses the inch columns from section 10.

````python
one = pitches[pitches["pitcher"] == 694973]
for pt, g in one.groupby("pitch_type"):
    plt.scatter(g["hb_in"], g["ivb_in"], s=8, label=pt)
plt.axhline(0, color="gray"); plt.axvline(0, color="gray")
plt.xlabel("Horizontal break (in)"); plt.ylabel("Induced vertical break (in)")
plt.legend(); plt.show()
````

**73. Velo by pitch of game**  
Tip: Careful: later pitch numbers include only pitchers who stayed in, mostly starters. That is a selection effect, not just fatigue.

````python
ff = pitches[pitches["pitch_type"] == "FF"]
ff.groupby("pitch_of_game")["release_speed"].mean().loc[1:100].plot()
plt.xlabel("Pitch of game"); plt.ylabel("Avg 4-seam velo (mph)")
plt.show()
````
